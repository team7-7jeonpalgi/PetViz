```python
df1.info()
```

```python
df1['성별'].value_counts(dropna=False)
```

```python
df1['성별'] = df1['성별'].replace({'암컷':"여", "수컷":"남"})
```

```python
gender_counts = df1['성별'].value_counts()
print(gender_counts)
```

```python
df1['종'] = df1['품종'].str.extract(r'\[(개|고양이)\]')
```

```python
df1['종'] = df1['종'].fillna('기타')
```

```python
species_counts = df1['종'].value_counts()
print(species_counts)
```

```python
df1['시도'] = df1['관할기관'].str.split().str[0]
df1['시도'] = df1['시도'].fillna('기타지역')
region_counts = df1['시도'].value_counts()
```

```python
print(region_counts)
```

```python
df1['중성화'].value_counts(dropna=False)
```

```python
df1['중성화']=df1['중성화'].fillna('미확인')
```

```python
df1['중성화'] = df1['중성화'].replace({'예':'o', '아니오':'x'})
```

```python
neuter_counts = df1['중성화'].value_counts()
print(neuter_counts)
```

```python
df1.drop(["상태", '발생장소','나이/체중','공고번호','보호센터연락처'], axis=1, inplace=True)
```

```python
df1.drop(["보호장소"], axis=1, inplace=True)
```
