```python
import pandas as pd
import glob
from natsort import natsorted
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
files = glob.glob('동물보호 업무부서 목록_*.xls')
recent_file = natsorted(seq=files, reverse=True)[0]
```

```python
def get_care_department():
    care_department = pd.read_excel(recent_file)
    care_department.drop(['번호', '전화번호'], axis=1, inplace=True)
    return care_department
```

```python
def save_care_department():
    get_care_department().to_csv(recent_file[:-4] + '.csv', encoding="utf-8-sig")
    return
```

```python
if __name__ == "__main__":
    save_care_department()
```
