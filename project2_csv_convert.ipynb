```python
import pandas as pd
import re
```

```python
df = pd.read_excel('입양대상 동물_20250516.xls')
df_hospital = pd.read_excel('동물병원 목록_20250516.xls')
df_shelter = pd.read_excel('동물보호센터 목록_20250516.xls')
```

## 전처리 함수

```python
def extract_type(x):
    return re.search(r'\[(.*?)\]', x).group(1) if '[' in x else '기타'

def extract_kind(x):
    return re.sub(r'\[.*?\]\s*', '', x)
```

## 입양대상_동물

```python
df = df[['공고번호', '품종', '성별', '접수일시', '관할기관', '보호센터']]
df.columns = ['announced_num', 'kind', 'sex', 'received_date', 'province', 'shelter']
```

```python
df['animal_type'] = df['kind'].apply(extract_type)
df['kind'] = df['kind'].apply(extract_kind)
```

```python
df = df[['received_date', 'animal_type', 'kind', 'sex', 'province', 'shelter']]
df.head()
```

## 동물 병원 목록

```python
df_hospital = df_hospital[['병원명', '소재지', '인허가번호']]
df_hospital.head()
```

```python
df_hospital.columns = ['hospital_name', 'location', 'id_code']
df_hospital.head()
```

## 동물보호센터 목록

```python
df_shelter = df_shelter[['관할구역', '보호센터명']]
df_shelter.head()
```

```python
df_shelter.columns = ['turf', 'center_name']
df_shelter.head()
```

```python
df.to_csv('abandoned_animal.csv', index=False, encoding='utf-8-sig')
df_hospital.to_csv('hospital.csv', index=False, encoding='utf-8-sig')
df_shelter.to_csv('shelter.csv', index=False, encoding='utf-8-sig')
```
